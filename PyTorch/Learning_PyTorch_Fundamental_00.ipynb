# PyTorch Linear Layer (`nn.Linear`) Revision Notes

### 1. Core Equation
$$y = x \cdot W^T + b$$
[cite: 10]
* Multi-dimensional version of $y = mx + b$[cite: 10].
* Under the hood, PyTorch executes an **Addmm** operation (Add + Matrix Multiply)[cite: 10].

---

### 2. Dimension Matching Rule
* Inner dimensions **must** match for valid matrix multiplication[cite: 10]:
  $$\underbrace{(M \times \mathbf{K})}_{\text{Input } x} \times \underbrace{(\mathbf{K} \times N)}_{\text{Transposed Weights } W^T} \longrightarrow \underbrace{(M \times N)}_{\text{Intermediate}} \xrightarrow{+ b} \underbrace{(M \times N)}_{\text{Output } y}$$
[cite: 10]

---

### 3. Parameters in `nn.Linear(in_features, out_features)`
* **`in_features`**: Must match the **last dimension** of your input $x$ ($K$)[cite: 10].
* **`out_features`**: The number of new features to project onto ($N$)[cite: 10].
* **Weights ($W$) shape**: `[out_features, in_features]` $\rightarrow$ Transposed ($W^T$) to `[in_features, out_features]` during computation[cite: 10].
* **Bias ($b$) shape**: `[out_features]` (added row-wise to offset outputs)[cite: 10].

---

### 4. Working Example Walkthrough
* **Input ($x = \text{tensor\_A}$):** Shape `[3, 2]` (3 samples, 2 features each)[cite: 10].
* **Layer:** `nn.Linear(in_features=2, out_features=6)`[cite: 10].
* **Computation:**
  $$(3 \times \mathbf{2}) \cdot (\mathbf{2} \times 6) + [6] \longrightarrow [3, 6]$$
[cite: 10]
* **Output:** Shape `torch.Size([3, 6])` (batch size remains `3`, feature dimension expands to `6`)[cite: 10].

---

### 5. Quick Debugging & Traps
* **`in_features` mismatch:** Setting `in_features != input.shape[-1]` causes `RuntimeError: mat1 and mat2 shapes cannot be multiplied`[cite: 4, 10].
* **`torch.manual_seed(n)`:** Fixes pseudo-random weight/bias initialization for 100% reproducible results[cite: 10].
* **`grad_fn=<AddmmBackward0>`:** PyTorch automatically tracks the operation history for automatic differentiation/backpropagation[cite: 4, 10].

## Reshaping, view, stacking, squeezing ,unsqueezing and permute 

### Reshaping 

* Reshaping (Rearranging the layout)

* What it is: Changing the grid layout without adding or removing any blocks.
* Lego Analogy: Taking your 24 blocks and shifting them from a flat 4×6 grid into a 3×8 grid or a 2×12 grid.
* Core Rule: The total number of blocks (24) never changes.


### View 

* View (The ultra-fast perspective)

* What it is: Creating a new "window" to look at the exact same blocks in memory. It is a faster, memory-safe version of reshaping.
* Lego Analogy: Looking at your 4×6 layout through a special stencil that forces you to see it as a 3×8 grid. Because you aren't actually touching or moving the physical blocks, it is instant—but it only works if your blocks are already sitting perfectly in a neat row in memory.
* PyTorch Code: tensor.view(3, 8)


### Stacking (Gluing layers together)

* What it is: Joining multiple arrays of the exact same shape along a brand new direction (axis).
* Lego Analogy: Taking three identical flat 4×6 grid layouts and stacking them directly on top of each other to build a 3D Lego house (a 3×4×6 block).
* PyTorch Code: torch.stack([t1, t2, t3], dim=0)

### Squeezing (Removing empty dimensions)

* What it is: Dropping a dimension that only has a size of 1. It gets rid of redundant "outer boxes."
* Lego Analogy: If you have a package containing exactly 1 flat grid of 4×6 legos, it acts like a 3D box (1×4×6). Squeezing it means throwing away the outer box and just treating it as a flat 2D grid (4×6).
* PyTorch Code: tensor.squeeze()

### Unsqueezing (Adding an empty dimension)

* What it is: The exact opposite of squeezing. It adds a brand new dimension with a size of 1.
* Lego Analogy: Taking your flat 2D grid (4×6) and putting it inside a single, flat shipping container. Now it is technically a 3D object (1×4×6) even though it's still just one layer thick.
* PyTorch Code: tensor.unsqueeze(dim=0)


### Permute / Transpose (Swapping the axes)

* What it is: Rotating or flipping the entire perspective of your data structure.
* Lego Analogy: Look at a 3D Lego structure (Length × Width × Height). Permute is like picking it up and flipping it on its side. What used to be the "Height" is now the "Length," and what used to be the "Length" is now the "Width." The blocks stay glued together; you are just changing your viewpoint.
* PyTorch Code: tensor.permute(2, 0, 1)

## Reproducibility (trying to take the random out of random)

### PyTorch Random Seeds 

* **The Problem:** Computers make "random" numbers for neural networks, so running the same code normally gives different results every time.
* **The Solution (Seed):** A seed is a starting code (like `42`) that forces the computer to pick the exact same "random" numbers every run.

#### The Rule to Remember:
A seed gets used up immediately. If you want two tensors to be identical, reset the seed right before making each one:

```python
torch.manual_seed(42)  # Set seed
a = torch.rand(2, 2)  # Uses the seed

torch.manual_seed(42)  # Reset seed
b = torch.rand(2, 2)  # Exactly matches 'a'

````python
random_tensor_without_seed_A = torch.rand(4,5)
random_tensor_without_seed_B = torch.rand(4,5)
random_tensor_without_seed_A == random_tensor_without_seed_B # as we can see that rand. takes all times different , so we need 
                                                             # to fix it by seeding it . 
````

````python
import torch 
import random 

Random_seed = 19

torch.manual_seed(Random_seed)
random_tensor_with_seed_A = torch.rand(4,5)

torch.manual_seed(Random_seed)
random_tensor_with_seed_B = torch.rand(4,5)

random_tensor_with_seed_A == random_tensor_with_seed_B
````

## PyTorch GPU Acceleration & Device Setup — Revision Notes

#### 1. Compute Devices
* **`cpu`**: Default hardware target on every computer; handles general sequential logic.
* **`cuda`**: NVIDIA platform used on cloud instances (e.g., Google Colab) and NVIDIA PCs.
* **`mps` (Metal Performance Shaders)**: Apple Silicon GPU engine used on M1/M2/M3 Macs.

---

#### 2. Device-Agnostic Setup Pattern
Write code that dynamically selects the best available accelerator[cite: 19, 21]:

```python
import torch

if torch.cuda.is_available():
    device = "cuda"  
elif torch.backends.mps.is_available():
    device = "mps"   
else:
    device = "cpu"   

print(f"Using device: {device}")

````python
import torch 
print(torch.backends.mps.is_available())

print(torch.backends.mps.is_built())
print(torch.cuda.is_available())
````

````python
import torch 

if torch.backends.mps.is_available() :
    device = "mps"
elif torch.cuda.is_available() :
    device = "cuda"
else :
    device = "cpu"        
print(f"the device using is {device}")    
````

````python
# Moving Tensors to GPU
tensor = torch.rand(4,5)
print(tensor.device) # cpu

# moving cpu to gpu 
tensor_gpu = tensor.to(device)
print(tensor_gpu.device)
````

````python
## Moving Back to CPU (For NumPy)
tensor_gpu = torch.rand(3,4).to(device)

## arr = tensor_gpu.numpy() Wrong approch 
arr = tensor_gpu.cpu().numpy()
print(arr)
print(type(arr))
````

````python
# Instead of creating on CPU then moving, create directly on device
tensor = torch.rand(3, 4, device=device)
print(tensor.device)   # mps:0  (created directly on GPU)
````

* Rule 1: Tensors on different devices CANNOT interact

        tensor_cpu + tensor_gpu → ERROR
        
        Solution: move both to same device first

* Rule 2: NumPy only works on CPU tensors

        Always: tensor.cpu().numpy()

* Rule 3: Never hardcode "mps" or "cpu"

        Always use: device = "mps" if ... else "cpu"

* Rule 4: Move model AND data to same device

        model.to(device) + X.to(device) + y.to(device)
