
### 10. Make a random tensor with shape `(1, 1, 1, 10)` and then create a new tensor with all the `1` dimensions removed to be left with a tensor of shape `(10)`. Set the seed to `7` when you create it and print out the first tensor and it's shape as well as the second tensor and it's shape.

The output should look like:

```
tensor([[[[0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297,
           0.3653, 0.8513]]]]) torch.Size([1, 1, 1, 10])
tensor([0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297, 0.3653,
        0.8513]) torch.Size([10])
```

````python
# Set seed
torch.manual_seed(7)



# Create random tensor
tensor_original = torch.rand(1,1,1,10)


# Remove single dimensions
torch.manual_seed(7)
tensor_squezed = tensor_original.squeeze()



# Print out tensors and their shapes
print(f"the original tensor without squeze is {tensor_original} with shape of {tensor_original.shape}")
print(f"The tensor after squezing one dimension is {tensor_squezed} with shape of {tensor_squezed.shape}")
````
